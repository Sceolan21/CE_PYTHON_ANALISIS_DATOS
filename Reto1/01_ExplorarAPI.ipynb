1. ¿Qué información proporciona el servicio?
Proporciona datos meteorológicos históricos diarios de la ubicación indicada: temperaturas media, máxima y mínima. También incluye metadatos como la zona horaria, altitud y unidades de medida.

2. ¿Qué significan latitude y longitude?
Son las coordenadas geográficas del lugar consultado:
- latitude=37.7796: latitud, posición norte-sur.
- longitude=-3.7849: longitud, posición este-oeste. El valor negativo indica oeste.
Estas coordenadas corresponden a una zona cercana a Jaén.

3. ¿Qué fechas se están consultando?
Desde el 1 de enero de 2025 hasta el 31 de enero de 2025, ambas fechas incluidas. En total se obtienen 31 registros diarios.

4. ¿Qué información solicitan temperature_2m_mean , temperature_2m_max y temperature_2m_min ?
- temperature_2m_mean: temperatura media diaria a dos metros del suelo.
- temperature_2m_max: temperatura máxima diaria a dos metros del suelo.
- temperature_2m_min: temperatura mínima diaria a dos metros del suelo.
Los valores se expresan en grados Celsius (°C).

5. ¿En qué formato se presenta la respuesta?
Se presenta en formato JSON. Contiene pares clave-valor y estructuras anidadas. Por ejemplo, dentro de daily aparecen listas con las fechas y temperaturas.

6. ¿Qué diferencias observas entre esa respuesta y una lista de Python?
La respuesta completa no es una simple lista, sino un objeto JSON parecido a un diccionario de Python:
- Tiene claves como latitude, timezone, daily_units y daily.
- Puede contener otros objetos y listas en su interior.
- Una lista de Python solo almacena elementos ordenados por posición.
- JSON es texto utilizado para intercambiar información; para trabajar con él en Python normalmente se convierte en diccionarios y listas.

7. ¿Qué ocurriría si cambiásemos end_date para consultar hasta el 15 de enero?
La consulta devolvería únicamente los datos comprendidos entre el 1 y el 15 de enero, ambas fechas incluidas. Las listas de fechas y temperaturas tendrían 15 elementos, por lo que la respuesta sería más pequeña.

8. ¿Qué ventaja tiene consultar una API frente a introducir los datos manualmente?
Permite obtener los datos de manera automática, rápida y estructurada. Además, reduce errores de transcripción, facilita actualizar la información y hace posible consultar distintos lugares o periodos cambiando solamente los parámetros de la URL.

## 6. Relaciona cada valor con su fecha

Las temperaturas están asociadas a fechas. Organizaremos los datos en un diccionario de registros que conserve la fecha y las temperaturas media, máxima y mínima.

Cada entrada tendrá esta estructura:

```python
fecha: {
    "media": media,
    "maxima": maxima,
    "minima": minima,
}
```

````python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
    "latitude": 37.7796,
    "longitude": -3.7849,
    "start_date": "2025-01-01",
    "end_date": "2025-01-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
}

respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()

fechas = datos["daily"]["time"]
temperaturas_medias = datos["daily"]["temperature_2m_mean"]
temperaturas_maximas = datos["daily"]["temperature_2m_max"]
temperaturas_minimas = datos["daily"]["temperature_2m_min"]

registros = {
    fecha: {"media": media, "maxima": maxima, "minima": minima}
    for fecha, media, maxima, minima in zip(
        fechas, temperaturas_medias, temperaturas_maximas, temperaturas_minimas
    )
}

print("Registros creados:", len(registros))
````

## Actividades

Comprueba que puedes mostrar un registro completo, localizar el día con la media más alta y calcular la amplitud térmica diaria (`máxima - mínima`).

### 1. Organiza los datos en un diccionario de registros

**Respuesta:**

El diccionario contiene 31 entradas, una por cada fecha. La fecha funciona como clave y su valor es otro diccionario con las temperaturas `media`, `maxima` y `minima`.

````python
print("Número de registros:", len(registros))
print(registros)
````

### 2. Muestra un registro completo

**Respuesta:**

El registro del `2025-01-01` contiene una media de `7.1 °C`, una máxima de `12.8 °C` y una mínima de `2.4 °C`.

````python
fecha_ejemplo = fechas[0]
registro_ejemplo = registros[fecha_ejemplo]

print("Fecha:", fecha_ejemplo)
print("Registro:", registro_ejemplo)
````

### 3. Localiza el día con la temperatura media más alta

**Respuesta:**

El día con la temperatura media más alta fue el `2025-01-27`, con `14.8 °C`. Ese día la máxima fue `16.8 °C` y la mínima `10.3 °C`.

````python
fecha_media_mas_alta = max(registros, key=lambda fecha: registros[fecha]["media"])
registro_media_mas_alta = registros[fecha_media_mas_alta]

print("Fecha con la media más alta:", fecha_media_mas_alta)
print("Registro:", registro_media_mas_alta)
````

### 4. Calcula la amplitud térmica diaria

**Respuesta:**

La amplitud térmica se obtiene restando la mínima a la máxima de cada fecha. La mayor amplitud fue de `13.4 °C` y se registró el `2025-01-26`. La amplitud media del periodo fue de `8.67 °C`.

````python
amplitudes_termicas = {
    fecha: round(registro["maxima"] - registro["minima"], 1)
    for fecha, registro in registros.items()
}

fecha_mayor_amplitud = max(amplitudes_termicas, key=amplitudes_termicas.get)
amplitud_media = sum(amplitudes_termicas.values()) / len(amplitudes_termicas)

print("Amplitudes térmicas:", amplitudes_termicas)
print("Fecha con mayor amplitud:", fecha_mayor_amplitud)
print("Mayor amplitud:", amplitudes_termicas[fecha_mayor_amplitud], "°C")
print("Amplitud media:", round(amplitud_media, 2), "°C")
````

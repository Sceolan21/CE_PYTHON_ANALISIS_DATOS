## 8. Estudia la evolución temporal

Analizaremos cómo cambiaron las temperaturas medias a lo largo de enero de 2025, conservando siempre sus fechas.

```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
    "latitude": 37.7796,
    "longitude": -3.7849,
    "start_date": "2025-01-01",
    "end_date": "2025-01-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
}

respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()

fechas = datos["daily"]["time"]
temperaturas = datos["daily"]["temperature_2m_mean"]

print("Días obtenidos:", len(fechas))
```

## Actividades

### 1. Muestra las temperaturas medias junto con sus fechas

**Respuesta:**

Se obtienen 31 pares `(fecha, temperatura)`, uno por cada día de enero. Conservar la fecha permite interpretar el orden temporal y localizar cada cambio.

```python
temperaturas_por_fecha = list(zip(fechas, temperaturas))

for fecha, temperatura in temperaturas_por_fecha:
    print(fecha, "->", temperatura, "°C")
```

### 2. Filtra la primera semana y la segunda quincena

**Respuesta:**

La primera semana comprende del `2025-01-01` al `2025-01-07` y contiene 7 valores. La segunda quincena comprende del `2025-01-16` al `2025-01-31` y contiene 16 valores.

```python
primera_semana = list(zip(fechas[:7], temperaturas[:7]))
segunda_quincena = list(zip(fechas[15:], temperaturas[15:]))

print("Primera semana:", primera_semana)
print("Segunda quincena:", segunda_quincena)
```

### 3. Calcula las variaciones entre días consecutivos e identifica el mayor aumento y descenso

**Respuesta:**

El mayor aumento fue de `5.3 °C`, entre el `2025-01-20` y el `2025-01-21`. El mayor descenso fue de `-5.7 °C`, entre el `2025-01-27` y el `2025-01-28`.

```python
variaciones = [
    {
        "desde": fechas[indice - 1],
        "hasta": fechas[indice],
        "variacion": round(temperaturas[indice] - temperaturas[indice - 1], 1),
    }
    for indice in range(1, len(temperaturas))
]

mayor_aumento = max(variaciones, key=lambda cambio: cambio["variacion"])
mayor_descenso = min(variaciones, key=lambda cambio: cambio["variacion"])

print("Variaciones:", variaciones)
print("Mayor aumento:", mayor_aumento)
print("Mayor descenso:", mayor_descenso)
```

### 4. Calcula una media móvil de tres días y compárala con los valores originales

**Respuesta:**

La media móvil comienza el `2025-01-03`, porque necesita tres observaciones. Produce 29 valores y suaviza los cambios bruscos: cada resultado resume el día actual y los dos anteriores.

```python
media_movil_3_dias = [
    round(sum(temperaturas[indice - 2:indice + 1]) / 3, 2)
    for indice in range(2, len(temperaturas))
]

comparacion_media_movil = [
    (fechas[indice], temperaturas[indice], media_movil_3_dias[indice - 2])
    for indice in range(2, len(temperaturas))
]

print("Fecha, valor original y media móvil:")
for fecha, original, movil in comparacion_media_movil:
    print(fecha, "-> original:", original, "°C | media móvil:", movil, "°C")
```

### 5. Divide el periodo en semanas y calcula la media de cada una

**Respuesta:**

- Semana 1 (`01–07`): `8.10 °C`.
- Semana 2 (`08–14`): `9.11 °C`.
- Semana 3 (`15–21`): `6.87 °C`.
- Semana 4 (`22–28`): `11.83 °C`.
- Semana 5 (`29–31`): `7.37 °C`.

La cuarta semana fue la más cálida y la tercera la más fría. La última semana solo contiene tres días, por lo que no es directamente comparable con las semanas completas.

```python
resumen_semanal = []

for inicio in range(0, len(temperaturas), 7):
    fin = min(inicio + 7, len(temperaturas))
    valores_semana = temperaturas[inicio:fin]
    resumen_semanal.append({
        "semana": len(resumen_semanal) + 1,
        "desde": fechas[inicio],
        "hasta": fechas[fin - 1],
        "dias": len(valores_semana),
        "media": round(sum(valores_semana) / len(valores_semana), 2),
    })

for semana in resumen_semanal:
    print(semana)
```

## Programa principal

A continuación se muestra el programa `meteorologia.py`, que obtiene los datos meteorológicos desde Open-Meteo.

````python
import requests
url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
"latitude": 37.7796,
"longitude": -3.7849,
"start_date": "2025-01-01",
"end_date": "2025-01-31",
"daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
"timezone": "Europe/Madrid",
}
respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()
print(datos)
````

## Funcionamiento del programa

> **Observa el resultado:**
>
> - `requests.get()` envía la petición HTTP.
> - `raise_for_status()` informa de errores HTTP.
> - `respuesta.json()` convierte la respuesta JSON en objetos de Python.
> - No es necesario importar `json`.

## Comprueba tu comprensión

### 1. ¿Para qué sirve `import requests`?

**Respuesta:**

Importa la biblioteca `requests`, que permite realizar peticiones HTTP desde Python, por ejemplo para consultar una API.

### 2. ¿Qué tipo de objeto es `parametros`?

**Respuesta:**

Es un diccionario de Python (`dict`). Sus claves representan los nombres de los parámetros y sus valores contienen los datos enviados a la API.

### 3. ¿Qué diferencia hay entre `url` y `parametros`?

**Respuesta:**

- `url` contiene la dirección base del servicio: <https://archive-api.open-meteo.com/v1/archive>.
- `parametros` contiene los detalles de la consulta, como las coordenadas, fechas, variables meteorológicas y zona horaria.
- `requests` utiliza ambos para construir la URL completa.

### 4. ¿Qué hacen `timeout=30` y `raise_for_status()`?

**Respuesta:**

- `timeout=30` evita que el programa espere indefinidamente si el servidor no responde. Si se supera el tiempo de espera, se genera una excepción.
- `raise_for_status()` comprueba el código de estado HTTP. Si la respuesta indica un error, como `404` o `500`, genera una excepción.

### 5. ¿Qué diferencia hay entre `respuesta` y `datos`?

**Respuesta:**

- `respuesta` es un objeto `Response` que contiene la respuesta HTTP completa: código de estado, cabeceras, contenido, etc.
- `datos` contiene el cuerpo JSON de la respuesta convertido a objetos de Python. En este caso, es principalmente un diccionario con otros diccionarios y listas en su interior.

### 6. ¿Qué podría ocurrir si falla la conexión o el servidor devuelve un error?

**Respuesta:**

El programa podría generar una excepción y detenerse:

- Si falla la conexión: `ConnectionError`.
- Si se supera el tiempo de espera: `Timeout`.
- Si el servidor devuelve un error HTTP: `HTTPError` al ejecutar `raise_for_status()`.

Estas excepciones podrían controlarse mediante `try` y `except`.

### 7. ¿Por qué separamos los parámetros de la dirección URL?

**Respuesta:**

Porque el código queda más claro, ordenado y fácil de modificar. Además, `requests` se encarga de construir y codificar correctamente la URL, evitando errores al concatenarla manualmente.

### 8. ¿En qué se diferencia una petición `GET` de una petición `POST`?

**Respuesta:**

- `GET` se utiliza normalmente para consultar u obtener información. Sus parámetros suelen enviarse en la URL y, en principio, no modifica los datos del servidor.
- `POST` se utiliza normalmente para enviar información al servidor, por ejemplo para crear un registro. Los datos suelen enviarse en el cuerpo de la petición.

En este programa se emplea `GET` porque únicamente se están consultando datos meteorológicos.

## 7. Analiza el periodo

Calcularemos medidas descriptivas para las temperaturas medias, máximas y mínimas de enero de 2025.

## Medidas descriptivas

- **Media aritmética:** suma de los valores dividida entre el número de observaciones.
- **Mediana:** valor central después de ordenar los datos.
- **Máximo y mínimo:** valores mayor y menor del conjunto.
- **Rango:** diferencia entre el máximo y el mínimo.
- **Varianza:** promedio de las diferencias al cuadrado respecto a la media. Se expresa en `°C²`.
- **Desviación típica:** raíz cuadrada de la varianza. Mide la dispersión en las mismas unidades que la temperatura.

> Como analizamos todos los días del periodo, utilizaremos la varianza y la desviación típica poblacionales.

```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
    "latitude": 37.7796,
    "longitude": -3.7849,
    "start_date": "2025-01-01",
    "end_date": "2025-01-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
}

respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()

temperaturas_medias = datos["daily"]["temperature_2m_mean"]
temperaturas_maximas = datos["daily"]["temperature_2m_max"]
temperaturas_minimas = datos["daily"]["temperature_2m_min"]

print("Valores por serie:", len(temperaturas_medias))
```

## Cálculo de las estadísticas

```python
from statistics import mean, median, pvariance, pstdev

series = {
    "Medias": temperaturas_medias,
    "Máximas": temperaturas_maximas,
    "Mínimas": temperaturas_minimas,
}

estadisticas = {
    nombre: {
        "media": mean(valores),
        "mediana": median(valores),
        "mínimo": min(valores),
        "máximo": max(valores),
        "rango": max(valores) - min(valores),
        "varianza": pvariance(valores),
        "desviación típica": pstdev(valores),
    }
    for nombre, valores in series.items()
}

for nombre, medidas in estadisticas.items():
    print(f"\nTEMPERATURAS {nombre.upper()}")
    for medida, valor in medidas.items():
        unidad = "°C²" if medida == "varianza" else "°C"
        print(f"{medida.capitalize()}: {valor:.2f} {unidad}")
```

## Resultados

|---|---:|---:|---:|---:|---:|---:|---:|

| Máximas | 13.34 °C | 12.90 °C | 9.10 °C | 18.60 °C | 9.50 °C | 6.37 °C² | 2.52 °C |
| Mínimas | 4.67 °C | 3.70 °C | -1.10 °C | 10.30 °C | 11.40 °C | 10.05 °C² | 3.17 °C |

## Conclusiones

- La temperatura mínima absoluta fue de `-1.10 °C`, lo que muestra que se produjeron condiciones de helada. La máxima absoluta alcanzó `18.60 °C`.
- Las temperaturas mínimas fueron las más variables: presentan el rango más amplio (`11.40 °C`) y la desviación típica más alta (`3.17 °C`).
- Las temperaturas máximas fueron las más estables, con el menor rango (`9.50 °C`) y la menor desviación típica (`2.52 °C`).
- Las desviaciones típicas indican que los valores suelen separarse de sus respectivas medias aproximadamente `2.79 °C` en las medias diarias, `2.52 °C` en las máximas y `3.17 °C` en las mínimas.

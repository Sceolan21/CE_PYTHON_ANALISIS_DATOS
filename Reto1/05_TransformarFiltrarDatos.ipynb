# Reto 1: Consulta y análisis de datos meteorológicos

## 5. Transforma y filtra datos

Utilizaremos comprensiones de listas para transformar y filtrar las temperaturas medias diarias.

La conversión de grados Celsius a Fahrenheit se calcula mediante:

$$F = \frac{9}{5}C + 32$$

```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
    "latitude": 37.7796,
    "longitude": -3.7849,
    "start_date": "2025-01-01",
    "end_date": "2025-01-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
}

respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()

temperaturas = datos["daily"]["temperature_2m_mean"]
print(temperaturas)
```

## Actividades

### 1. Convierte las temperaturas a Fahrenheit

**Respuesta:**

Se aplica la fórmula a cada temperatura mediante una comprensión de lista. Los resultados se redondean a un decimal.

```python
temperaturas_fahrenheit = [round(temperatura * 9 / 5 + 32, 1) for temperatura in temperaturas]
print(temperaturas_fahrenheit)
```

### 2. Separa las temperaturas según la media del periodo

Crea una lista con las temperaturas menores o iguales a la media y otra con las temperaturas superiores a ella.

**Respuesta:**

La media del periodo es `8.82 °C`. Hay 17 temperaturas menores o iguales a la media y 14 temperaturas superiores.

```python
media_periodo = sum(temperaturas) / len(temperaturas)
temperaturas_hasta_media = [temperatura for temperatura in temperaturas if temperatura <= media_periodo]
temperaturas_sobre_media = [temperatura for temperatura in temperaturas if temperatura > media_periodo]

print("Media del periodo:", round(media_periodo, 2), "°C")
print("Menores o iguales a la media:", temperaturas_hasta_media)
print("Superiores a la media:", temperaturas_sobre_media)
```

### 3. Clasifica cada temperatura como inferior, igual o superior a 15 °C

**Respuesta:**

Las 31 temperaturas medias son inferiores a `15 °C`. No hay ninguna igual ni superior a ese valor.

```python
clasificaciones_15 = [
    (temperatura, "inferior" if temperatura < 15 else "igual" if temperatura == 15 else "superior")
    for temperatura in temperaturas
]

print(clasificaciones_15)
```

### 4. Crea una lista con los índices de los días cuya temperatura media superó los 15 °C

**Respuesta:**

La lista está vacía (`[]`) porque ninguna temperatura media superó los `15 °C`. Los índices de Python comienzan en cero.

```python
indices_superiores_15 = [
    indice for indice, temperatura in enumerate(temperaturas) if temperatura > 15
]

print(indices_superiores_15)
```

### 5. Construye tuplas con cada temperatura en Celsius y su equivalente en Fahrenheit

**Respuesta:**

Cada tupla contiene primero la temperatura original en Celsius y después su conversión a Fahrenheit redondeada a un decimal.

```python
temperaturas_celsius_fahrenheit = [
    (temperatura, round(temperatura * 9 / 5 + 32, 1))
    for temperatura in temperaturas
]

print(temperaturas_celsius_fahrenheit)
```

### 6. Calcula la variación respecto al día anterior

**Respuesta:**

Se obtienen 30 variaciones porque el primer día no tiene un día anterior con el que compararse. Un valor positivo representa un aumento y uno negativo representa un descenso.

```python
variaciones = [
    round(temperaturas[indice] - temperaturas[indice - 1], 1)
    for indice in range(1, len(temperaturas))
]

print(variaciones)
```

### 7. Define umbrales para clasificar los días como fríos, templados o cálidos

**Criterio elegido:**

- **Frío:** temperatura media inferior a `7 °C`.
- **Templado:** temperatura media desde `7 °C` hasta menos de `12 °C`.
- **Cálido:** temperatura media igual o superior a `12 °C`.

Estos umbrales permiten dividir de forma sencilla el intervalo observado en el periodo; no representan una clasificación meteorológica universal.

**Respuesta:**

Con este criterio hay 8 días fríos, 17 templados y 6 cálidos.

```python
clasificacion_dias = [
    (temperatura, "frío" if temperatura < 7 else "templado" if temperatura < 12 else "cálido")
    for temperatura in temperaturas
]

dias_frios = sum(clasificacion == "frío" for _, clasificacion in clasificacion_dias)
dias_templados = sum(clasificacion == "templado" for _, clasificacion in clasificacion_dias)
dias_calidos = sum(clasificacion == "cálido" for _, clasificacion in clasificacion_dias)

print(clasificacion_dias)
print("Días fríos:", dias_frios)
print("Días templados:", dias_templados)
print("Días cálidos:", dias_calidos)
```

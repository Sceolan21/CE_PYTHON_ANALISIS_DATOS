# Práctica: entornos virtuales con `venv`

## Objetivo

Crearemos un entorno virtual aislado, instalaremos bibliotecas, comprobaremos el intérprete utilizado, generaremos `requirements.txt` y ejecutaremos un pequeño programa de análisis de temperaturas. Toda la práctica está guardada dentro de la carpeta `venv`.

## 1. ¿Qué es un entorno virtual?

Un entorno virtual es un espacio aislado que utiliza su propio intérprete de Python y sus propias bibliotecas. Permite que cada proyecto tenga dependencias y versiones diferentes sin modificar la instalación global.

Sus principales ventajas son:

- Evitar conflictos entre proyectos.
- No instalar bibliotecas en el Python global.
- Registrar versiones concretas de las dependencias.
- Facilitar la reproducción del proyecto en otro equipo.

## 2. Estructura de la práctica

```text
venv/
├── .venv/
├── resultados/
│   ├── temperaturas_practica.csv
│   └── temperaturas_practica.png
├── .gitignore
├── comprobar_entorno.py
├── estadisticas.py
├── Practica_Entornos_Virtuales.ipynb
└── requirements.txt
```

La carpeta de la práctica se llama `venv`, mientras que `.venv` es el entorno virtual real. El punto inicial permite distinguir ambos conceptos.

## 3. Creación y activación

Desde PowerShell y dentro de la carpeta de la práctica se ejecutan estos comandos:

```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
```

Cuando está activo, la terminal muestra normalmente `(.venv)` delante del indicador. Cada terminal nueva necesita activar otra vez el entorno. Para salir se utiliza:

```powershell
deactivate
```

## 4. Comprobación del intérprete

`sys.executable` muestra el Python que ejecuta el código. Si `sys.prefix` y `sys.base_prefix` son diferentes, estamos dentro de un entorno virtual.

````python
import sys

print("Ejecutable:", sys.executable)
print("Prefijo actual:", sys.prefix)
print("Prefijo base:", sys.base_prefix)
print("Estamos en un entorno virtual:", sys.prefix != sys.base_prefix)
````

## 5. Instalación de bibliotecas

Con el entorno activado se actualizó `pip` y se instalaron las tres bibliotecas de la actividad:

```powershell
python -m pip install --upgrade pip
python -m pip install numpy pandas matplotlib
```

Se utiliza `python -m pip` para garantizar que `pip` pertenece al mismo intérprete que ejecuta el programa.

````python
import matplotlib
import numpy as np
import pandas as pd

print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
````

## 6. Programa de análisis

`estadisticas.py` utiliza las tres bibliotecas instaladas:

- NumPy calcula la media.
- pandas organiza las fechas y temperaturas en un `DataFrame` y genera el CSV.
- Matplotlib crea una gráfica de líneas en PNG.

El programa se ejecuta con el Python del entorno:

```powershell
python estadisticas.py
```

````python
temperaturas = np.array([7.1, 7.3, 7.4, 8.1, 9.5, 8.7, 8.6])
datos = pd.DataFrame({"temperatura_media": temperaturas})

print(datos)
print(f"Temperatura media: {temperaturas.mean():.2f} °C")
````

### Resultado

El programa ha calculado una temperatura media de `8.10 °C` y ha generado correctamente:

- `resultados/temperaturas_practica.csv`.
- `resultados/temperaturas_practica.png`.

## 7. Registro de dependencias

Las versiones instaladas se guardaron mediante:

```powershell
python -m pip freeze > requirements.txt
```

`requirements.txt` incluye también dependencias indirectas necesarias para que NumPy, pandas y Matplotlib funcionen.

## 8. Reconstrucción en otro equipo

No se debe copiar ni compartir `.venv`. Otro usuario puede reconstruir el entorno con:

```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install -r requirements.txt
```

El código fuente y `requirements.txt` son suficientes para reproducir las dependencias.

## 9. Archivos excluidos de Git

`.gitignore` excluye el entorno y los archivos temporales:

```gitignore
.venv/
__pycache__/
*.pyc
.ipynb_checkpoints/
.env
```

`.venv` puede ocupar mucho espacio y depende del sistema operativo. Por eso se reconstruye y no se guarda en el repositorio.

## 10. Diferencia entre `.venv` y `.env`

- `.venv` contiene el intérprete y las bibliotecas del entorno virtual.
- `.env` almacena variables de configuración, como claves o direcciones de servicios.

Aunque los nombres se parecen, tienen finalidades diferentes. Esta práctica utiliza `.venv` y no necesita un archivo `.env`.

### 1. ¿Qué cambia al activar el entorno?

**Respuesta:**

Se modifica temporalmente el `PATH` de la terminal. Los comandos `python` y `pip` pasan a utilizar los ejecutables de `.venv`. Normalmente aparece también `(.venv)` delante del indicador de la terminal.

### 2. ¿Hay que enviar `.venv` al compartir el proyecto?

**Respuesta:**

No. Se comparte el código fuente y `requirements.txt`. La otra persona crea su propio entorno e instala las dependencias con `python -m pip install -r requirements.txt`.

### 3. ¿Por qué se recomienda utilizar `python -m pip`?

**Respuesta:**

Porque ejecuta el módulo `pip` asociado al intérprete seleccionado. Así se evita instalar una biblioteca en un Python diferente del que posteriormente ejecutará el programa.

## 4. Calcula estadísticas con las temperaturas

En este apartado utilizaremos la lista de temperaturas medias diarias para calcular estadísticas y practicar operaciones con listas.

```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
    "latitude": 37.7796,
    "longitude": -3.7849,
    "start_date": "2025-01-01",
    "end_date": "2025-01-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
}

respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()

temperaturas = datos["daily"]["temperature_2m_mean"]
temperaturas_minimas = datos["daily"]["temperature_2m_min"]
temperaturas_maximas = datos["daily"]["temperature_2m_max"]

print("Temperaturas medias:", temperaturas)
print("Temperaturas mínimas:", temperaturas_minimas)
print("Temperaturas máximas:", temperaturas_maximas)
```

## Estadísticas básicas

Calculamos la suma, la media, el valor máximo y el valor mínimo de las temperaturas.

```python
suma_temperaturas = sum(temperaturas)
media_temperaturas = suma_temperaturas / len(temperaturas)
temperatura_maxima = max(temperaturas)
temperatura_minima = min(temperaturas)

suma_minimas = sum(temperaturas_minimas)
media_minimas = suma_minimas / len(temperaturas_minimas)
maxima_de_minimas = max(temperaturas_minimas)
minima_de_minimas = min(temperaturas_minimas)

suma_maximas = sum(temperaturas_maximas)
media_maximas = suma_maximas / len(temperaturas_maximas)
maxima_de_maximas = max(temperaturas_maximas)
minima_de_maximas = min(temperaturas_maximas)

print("TEMPERATURAS MEDIAS")
print("Suma:", suma_temperaturas, "°C")
print("Media:", round(media_temperaturas, 2), "°C")
print("Máxima:", temperatura_maxima, "°C")
print("Mínima:", temperatura_minima, "°C")

print("\nTEMPERATURAS MÍNIMAS")
print("Suma:", suma_minimas, "°C")
print("Media:", round(media_minimas, 2), "°C")
print("Máxima:", maxima_de_minimas, "°C")
print("Mínima:", minima_de_minimas, "°C")

print("\nTEMPERATURAS MÁXIMAS")
print("Suma:", suma_maximas, "°C")
print("Media:", round(media_maximas, 2), "°C")
print("Máxima:", maxima_de_maximas, "°C")
print("Mínima:", minima_de_maximas, "°C")
```

### 5. Crea una lista de máximas superiores a 15 °C y otra de mínimas inferiores a 10 °C

**Respuesta:**

La lista de temperaturas máximas superiores a `15 °C` contiene 9 valores. La lista de temperaturas mínimas inferiores a `10 °C` contiene 29 valores.

```python
maximas_superiores_15 = [temperatura for temperatura in temperaturas_maximas if temperatura > 15]
minimas_inferiores_10 = [temperatura for temperatura in temperaturas_minimas if temperatura < 10]

print("Máximas superiores a 15 °C:", maximas_superiores_15)
print("Mínimas inferiores a 10 °C:", minimas_inferiores_10)
```

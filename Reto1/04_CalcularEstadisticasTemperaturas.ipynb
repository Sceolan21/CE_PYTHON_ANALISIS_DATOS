## 4. Calcula estadísticas con las temperaturas

En este apartado utilizaremos la lista de temperaturas medias diarias para calcular estadísticas y practicar operaciones con listas.

```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
parametros = {
    "latitude": 37.7796,
    "longitude": -3.7849,
    "start_date": "2025-01-01",
    "end_date": "2025-01-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
}

respuesta = requests.get(url, params=parametros, timeout=30)
respuesta.raise_for_status()
datos = respuesta.json()

temperaturas = datos["daily"]["temperature_2m_mean"]
print(temperaturas)
```

## Estadísticas básicas

Calculamos la suma, la media, el valor máximo y el valor mínimo de las temperaturas.

```python
suma_temperaturas = sum(temperaturas)
media_temperaturas = suma_temperaturas / len(temperaturas)
temperatura_maxima = max(temperaturas)
temperatura_minima = min(temperaturas)

print("Suma:", suma_temperaturas, "°C")
print("Media:", round(media_temperaturas, 2), "°C")
print("Máxima:", temperatura_maxima, "°C")
print("Mínima:", temperatura_minima, "°C")
```

## Comprueba tu comprensión

### 1. ¿Cuáles son las cinco primeras y las cinco últimas temperaturas?

**Respuesta:**

Las cinco primeras son `[7.1, 7.3, 7.4, 8.1, 9.5]` y las cinco últimas son `[14.8, 9.1, 8.1, 7.4, 6.6]`.

### 3. ¿Cuál es la diferencia entre la temperatura más alta y la más baja?

**Respuesta:**

La diferencia es `10.5 °C`, obtenida al restar `4.3 °C` a `14.8 °C`.

```python
diferencia = max(temperaturas) - min(temperaturas)
print("Diferencia entre máxima y mínima:", diferencia, "°C")
```

### 4. ¿Cuántos días superaron los 15 °C? ¿Cuántos estuvieron por debajo de 10 °C?

**Respuesta:**

Ningún día tuvo una temperatura media superior a `15 °C` y 19 días estuvieron por debajo de `10 °C`.

```python
dias_superiores_15 = sum(temperatura > 15 for temperatura in temperaturas)
dias_inferiores_10 = sum(temperatura < 10 for temperatura in temperaturas)

print("Días con más de 15 °C:", dias_superiores_15)
print("Días con menos de 10 °C:", dias_inferiores_10)
```

### 5. Crea una lista con las temperaturas superiores a 15 °C y otra con las inferiores a 10 °C

**Respuesta:**

La lista de temperaturas superiores a `15 °C` está vacía. La lista de temperaturas inferiores a `10 °C` contiene 19 valores.

```python
temperaturas_superiores_15 = [temperatura for temperatura in temperaturas if temperatura > 15]
temperaturas_inferiores_10 = [temperatura for temperatura in temperaturas if temperatura < 10]

print("Superiores a 15 °C:", temperaturas_superiores_15)
print("Inferiores a 10 °C:", temperaturas_inferiores_10)
```

### 6. Comprueba si todas las temperaturas son superiores a 0 °C y si alguna supera los 20 °C

**Respuesta:**

Todas las temperaturas medias son superiores a `0 °C`, pero ninguna supera los `20 °C`.

```python
todas_superiores_0 = all(temperatura > 0 for temperatura in temperaturas)
alguna_superior_20 = any(temperatura > 20 for temperatura in temperaturas)

print("Todas son superiores a 0 °C:", todas_superiores_0)
print("Alguna supera los 20 °C:", alguna_superior_20)
```

### 7. Redondea los valores a un decimal y guarda el resultado en otra lista

**Respuesta:**

Los valores ya tienen un decimal, por lo que la nueva lista contiene los mismos valores que la original.

```python
temperaturas_redondeadas = [round(temperatura, 1) for temperatura in temperaturas]
print(temperaturas_redondeadas)
```

### 8. Ordena las temperaturas de menor a mayor sin alterar la lista original

**Respuesta:**

La función `sorted()` devuelve una lista nueva ordenada y conserva `temperaturas` sin modificaciones.

```python
temperaturas_ordenadas = sorted(temperaturas)

print("Lista ordenada:", temperaturas_ordenadas)
print("Lista original:", temperaturas)
```
